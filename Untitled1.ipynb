```python
def generate_html_structure():
    html_content = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Jewelsfield Nursery and Primary School</title>
</head>
<body>
    <header>
        <h1>Jewelsfield Nursery and Primary School</h1>
        <p>Nurturing young minds for a brighter future</p>
    </header>
    <nav>
        <ul>
            <li><a href="#home">Home</a></li>
            <li><a href="#about">About Us</a></li>
            <li><a href="#admissions">Admissions</a></li>
            <li><a href="#gallery">Gallery</a></li>
            <li><a href="#contact">Contact Us</a></li>
        </ul>
    </nav>
    <main>
        <section id="home" class="hero">
            <h2>Welcome to Jewelsfield!</h2>
            <p>Where learning is an adventure.</p>
        </section>
        <section id="about" class="about-us">
            <h2>About Our School</h2>
            <p>Dedicated to providing quality education...</p>
        </section>
        <section id="admissions" class="admissions">
            <h2>Admissions</h2>
            <p>Information on how to enroll your child.</p>
        </section>
        <section id="gallery" class="gallery">
            <h2>Our Gallery</h2>
            <p>A glimpse into school life.</p>
        </section>
        <section id="contact" class="contact">
            <h2>Contact Us</h2>
            <p>Reach out to us for more information.</p>
        </section>
    </main>
    <footer>
        <p>&copy; 2023 Jewelsfield Nursery and Primary School. All rights reserved.</p>
        <p>Follow us on social media!</p>
    </footer>
</body>
</html>
    """
    return html_content

# Print the generated HTML to verify
print(generate_html_structure())
```

```python
def generate_css_styles():
    css_content = """
body {
    font-family: Arial, sans-serif;
    margin: 0;
    line-height: 1.6;
    color: #333;
    background-color: #f4f4f4;
}

header {
    background: #0056b3;
    color: #fff;
    padding: 1rem 0;
    text-align: center;
}

header h1 {
    margin: 0;
    font-size: 2.5rem;
}

header p {
    font-size: 1.2rem;
}

nav {
    background: #004085;
    color: #fff;
    padding: 0.5rem 0;
}

nav ul {
    list-style: none;
    padding: 0;
    margin: 0;
    display: flex;
    justify-content: center;
    flex-wrap: wrap; /* Allow navigation items to wrap on smaller screens */
}

nav ul li a {
    color: #fff;
    text-decoration: none;
    padding: 0.75rem 1.5rem;
    display: block;
    transition: background-color 0.3s ease;
}

nav ul li a:hover {
    background-color: #002752;
}

main {
    padding: 20px;
    max-width: 1200px;
    margin: 20px auto;
    background: #fff;
    box-shadow: 0 0 10px rgba(0, 0, 0, 0.1);
}

section {
    padding: 2rem;
    margin-bottom: 20px;
    background-color: #f9f9f9;
    border-radius: 8px;
}

section h2 {
    color: #0056b3;
    font-size: 2rem;
    margin-bottom: 1rem;
}

footer {
    background: #333;
    color: #fff;
    text-align: center;
    padding: 1rem 0;
    margin-top: 20px;
}

footer p {
    margin: 0.5rem 0;
}

/* Responsive Design */

/* Tablet styles */
@media (max-width: 768px) {
    header h1 {
        font-size: 2rem;
    }

    header p {
        font-size: 1rem;
    }

    nav ul {
        flex-direction: column;
        align-items: center;
    }

    nav ul li a {
        padding: 0.5rem 1rem;
        width: 100%;
        text-align: center;
    }

    main {
        margin: 10px auto;
        padding: 15px;
    }

    section {
        padding: 1.5rem;
    }

    section h2 {
        font-size: 1.75rem;
    }
}

/* Mobile styles */
@media (max-width: 480px) {
    body {
        font-size: 14px;
    }

    header {
        padding: 0.75rem 0;
    }

    header h1 {
        font-size: 1.8rem;
    }

    header p {
        font-size: 0.9rem;
    }

    nav ul li a {
        font-size: 0.9rem;
        padding: 0.4rem 0.8rem;
    }

    main {
        margin: 5px auto;
        padding: 10px;
    }

    section {
        padding: 1rem;
    }

    section h2 {
        font-size: 1.5rem;
    }

    footer {
        font-size: 0.8rem;
    }
}
    """
    return css_content

# Print the generated CSS to verify
print(generate_css_styles())
```

```python
html_content = generate_html_structure()
css_content = generate_css_styles()

# Embed CSS directly into the HTML head
# Find the closing </head> tag and insert the CSS before it
closing_head_tag = '</head>'
style_tag = f'<style>\n{css_content}\n</style>\n'

# Find the index to insert the style tag
insert_index = html_content.find(closing_head_tag)

if insert_index != -1:
    # Insert the style tag just before the closing </head> tag
    final_html_content = html_content[:insert_index] + style_tag + html_content[insert_index:]
else:
    # Fallback if </head> is not found, though it should be in our generated HTML
    print("Warning: </head> tag not found, appending CSS to the end of HTML.")
    final_html_content = html_content + style_tag

# Write the combined HTML to index.html
with open('index.html', 'w') as f:
    f.write(final_html_content)

print("index.html created successfully with embedded CSS.")
```
